```python
import urllib.request
import torch
url = "https://raw.githubusercontent.com/karpathy/makemore/master/names.txt"

urllib.request.urlretrieve(url, "names.txt")
```

```python
words = open('names.txt' , 'r').read().splitlines()
words[:10]
```

```python
chars = sorted(list(set(''.join(words))))      # list of all unique characters
# chars
stoi = {s:i+1 for i , s in enumerate(chars)}
# stoi
stoi['.'] = 0
stoi
itos = {i:s for s,i in stoi.items()}
itos
```

```python
# creating set of bigram of training model (x = input , y = target)
xs , ys = [] , []
for w in words[:1]:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
        print(ch1 , ch2)
        xs.append(row)
        ys.append(col)

xs = torch.tensor(xs)
ys = torch.tensor(ys)   
```

```python
# we can not feeding this integer to the neural network so we create float vector of this integer
# for that we will use One Hot Encoding function
import torch.nn.functional as F
x_encode = F.one_hot(xs , num_classes = 27).float()
x_encode
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.imshow(x_encode)
```

```python
# The Neural network : one linear layer of neurons implemented with matrix multiplication
W = torch.randn(27,1)
x_encode @ W     # input * weights
```

```python
W      # only for one neuron
```

```python
# for multiple neurons (randomly initializes 27 neuron's weights, each neuron receives 27 inputs)
g = torch.Generator().manual_seed(2147483647)
W = torch.randn(27 ,27 , generator = g)
x_encode @ W
```

```python
# (5 , 27) @ (27 , 27)  = (5 ,27) which means 5 inputs and 27 output of neuron which are input for next layer
```

```python
# for evaluate count matrix we have to find exponent of all these (5 ,27) 
logits = x_encode @ W
counts = logits.exp()        #equivalent to N
counts
```

```python
# transforming neural nets output into probabilities (softmax)
probs = counts / counts.sum(1 , keepdim = True)
probs     # probabilities for the next character
```

```python
# find out loss function as average negative log likelihood
nlls = torch.zeros(5)
for i in range(5):
    x = xs[i].item();    # input character index
    y = ys[i].item();    # output(label) character index
    print('-------')
    print(f'Bigram example {i+1}: {itos[x]}{itos[y]} (indexes {x}{y})')
    print(f'input to the neural network:', x)
    print(f'output probabilities from the neural network:', probs[i])
    print(f'label (actual next character):', y)
    p = probs[i , y]
    print(f'probabilites assigned by the network to the correct character:', p.item())
    logp = torch.log(p)
    print(f'log likelihood:' , logp.item())
    nll = -logp
    print(f'negative log likelihood:', nll.item())
    nlls[i] = nll
    
print('=======')
print(f'Average negative log likelihood i.e. loss =', nlls.mean().item())
```

```python
# Decreasing loss by doing iteration of forward pass , backward pass and update

g = torch.Generator().manual_seed(2147483647)
W = torch.randn(27 ,27 , generator = g , requires_grad = True)

for k in range(10):

    # forward pass
    x_encode = F.one_hot(xs , num_classes = 27).float()
    logits = x_encode @ W
    counts = logits.exp()        #equivalent to N
    probs = counts / counts.sum(1 , keepdim = True)
    loss = -probs[torch.arange(5) , ys].log().mean()
    print(loss.item())
    
    # backward pass
    W.grad = None
    loss.backward()
    
    # update
    W.data += -0.1 * W.grad
```

```python
# Putting everything together
# creating set of bigram of training model (x = input , y = target)
xs , ys = [] , []
for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
#         print(ch1 , ch2)
        xs.append(row)
        ys.append(col)
xs = torch.tensor(xs)
ys = torch.tensor(ys)  
number = xs.nelement()
print('-----')
print(f'number of examples : ', number)
print('-----')
# Decreasing loss by doing iteration of forward pass , backward pass and update

g = torch.Generator().manual_seed(2147483647)
W = torch.randn(27 ,27 , generator = g , requires_grad = True)

for k in range(100):

    # forward pass
    x_encode = F.one_hot(xs , num_classes = 27).float()
    logits = x_encode @ W
    counts = logits.exp()        #equivalent to N
    probs = counts / counts.sum(1 , keepdim = True)
    loss = -probs[torch.arange(number) , ys].log().mean()
    print(loss.item())
    
    # backward pass
    W.grad = None
    loss.backward()
    
    # update
    W.data += -50 * W.grad
```
