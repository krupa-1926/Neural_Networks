```python
N = torch.zeros((28, 28) , dtype = torch.int32)
```

```python
# we have to convert string (character) into integer as our array is int
chars = sorted(list(set(''.join(words))))      # list of all unique characters
# chars
stoi = {s:i for i , s in enumerate(chars)}
# stoi
stoi['<S>'] = 26
stoi['<E>'] = 27
stoi
```

```python
for w in words:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
        N[row , col] += 1 
```

```python
# as in the above figure you can see that <E>a to <E><E> is always Zero as <E> is never going to be first character of bigram
# same as a<S> to <S><E> is always zero as <S> is never going to be last character of bigram 
# so deleting spurious <S> , <E> tokens in a favors of a single "." token
N = torch.zeros((27, 27) , dtype = torch.int32)
```

```python
# we have to convert string (character) into integer as our array is int
chars = sorted(list(set(''.join(words))))      # list of all unique characters
# chars
stoi = {s:i+1 for i , s in enumerate(chars)}
# stoi
stoi['.'] = 0
stoi
itos = {i:s for s,i in stoi.items()}
itos
```

```python
for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
        N[row , col] += 1 
```

```python
# the above result still looks little ugly so make it more visual
import matplotlib.pyplot as plt
%matplotlib inline

plt.figure(figsize = (16,16))
plt.imshow(N , cmap = "Blues")

for i in range(27):
    for j in range(27):
        chstr = itos[i] + itos[j]
        plt.text(j , i , chstr, ha="center" , va="bottom" , color="gray")
        plt.text(j , i, N[i , j].item(), ha="center" , va="top" , color="gray")
plt.axis("off");
```

```python
# now let's create list of names
g = torch.Generator().manual_seed(2147483647)

for i in range(20):
    out = []
    ix = 0
    while True:
        p = N[ix].float()
        p = p / p.sum()
        ix = torch.multinomial(p , num_samples = 1 , replacement = True , generator = g).item()
        out.append(itos[ix])
        if(ix == 0):
            break
    print(''.join(out)) 
```

```python
P = N.float()
P = P.sum(1, keepdim = True).shape
P

# 27 , 27 which is N
# 27 , 1  which is P both are broadcastable
```

```python
P = N.float()
P = P.sum(1).shape
P

# 27 , 27 which is N
# 27   which is P both are broadcastable as now P becomes 1 ,27 but it can change the result
```

```python
# Efficiency : vectorized normalization of rows (broadcasting)
# what we are doing here is we are always fetching row of N from the counnt matrix , converting into float and dividing ans we are doing this again and again for every iteration
# it is extremely inefficient and wasteful  [p = N[ix].float();  p = p / p.sum()]
P = N.float()
P = P / P.sum(1, keepdim = True)
```

```python
# now let's create list of names with more efficiency
g = torch.Generator().manual_seed(2147483647)

for i in range(20):
    out = []
    ix = 0
    while True:
        p = P[ix]
        ix = torch.multinomial(p , num_samples = 1 , replacement = True , generator = g).item()
        out.append(itos[ix])
        if(ix == 0):
            break
    print(''.join(out)) 
```

```python
# GOAL : Maximize likelihood of data with respect to model parameters 
# equivalent to maximize the log likelihood (beacuse log is monotonic)
# equivalent to minimize the negative log likelihood
# equivalent to minimize the average negative log likelihood 

# log(a*b*c) = log(a) + log(b) + log(c)
```

```python
# Quality of the model (loss function = average negative log likelihood )
# (loss function is lower, better quality of model)

log_likelihood = 0.0
n = 0

for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
        prob = P[row , col]
        logprob = torch.log(prob)
        log_likelihood += logprob
        n += 1
#         print(f'{ch1}{ch2}: {prob:.4f} {logprob:.4f}')

print('---')
print(f'{log_likelihood=})')
negate_log_likelihood = -log_likelihood
print(f'{negate_log_likelihood=}')
print(f'{negate_log_likelihood/n}')
```

```python
# in the above example you can see loss is infinity as j and q has 0 probability
# to make the model smoother, we can add fake counts so that loss is not going to be infinity
P = (N+1).float()
P /= P.sum(1 , keepdim = True)
```

```python
# now you can see loss is not infinity because j and q has very small probability
log_likelihood = 0.0
n = 0

for w in ["krupjq"]:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs, chs[1:]):
        row = stoi[ch1]
        col = stoi[ch2]
        prob = P[row , col]
        logprob = torch.log(prob)
        log_likelihood += logprob
        n += 1
        print(f'{ch1}{ch2}: {prob:.4f} {logprob:.4f}')

print('---')
print(f'{log_likelihood=})')
negate_log_likelihood = -log_likelihood
print(f'{negate_log_likelihood=}')
print(f'{negate_log_likelihood/n}')
```
